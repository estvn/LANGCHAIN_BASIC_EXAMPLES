```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_openai import ChatOpenAI

load_dotenv(find_dotenv(usecwd=True))

llm = ChatOpenAI(
    model="stealth/crouton-alpha",
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"],
)
```

```python
# llm.invoke("La profesion de ingeniero de IA tiene un futuro prometedor?")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system",
     "Solo respondés preguntas sobre el futuro de la IA. "
     "Si la pregunta es de otro tema, respondé exactamente: "
     "'Solo puedo ayudarte con el futuro de la IA.'"),
    ("human", "{question}"),
])
```

```python
chain = prompt | llm
```

```python
chain.invoke({"question": "¿Como te llamas? de que color es el cielo?"})
```
